```python
df = df[
    ['Country', 'Region', 'State',
     'latitude', 'longitude',
     'Year',
     'Killed', 'Wounded',
     'Attack_Type',
     'Target_Type',
     'Group_Name',
     'Weap_Type']
]
```

```python
df = df[(df['latitude'].between(-90, 90)) &
        (df['longitude'].between(-180, 180))]
```

```python
df_country = df.copy()
```

```python
df_country[['Killed','Wounded']] = df_country[['Killed','Wounded']].fillna(0)
```

```python
df_country['Fatalities'] = df_country['Killed'] + df_country['Wounded']
```

```python
country_agg = (
    df_country
    .groupby(['Country'])
    .agg(
        Attacks=('Year', 'count'),
        Killed=('Killed', 'sum'),
        Wounded=('Wounded', 'sum'),
        Fatalities=('Fatalities', 'sum')
    )
    .reset_index()
)
```

```python
country_agg['country_id'] = country_agg['Country'].astype('category').cat.codes
```

```python
country_agg
```

```python
import geopandas as gpd

world = gpd.read_file("ne_50m_admin_0_countries/ne_50m_admin_0_countries.shp")
```

```python
world = world.merge(
    country_agg,
    left_on='NAME',        # shapefile country name
    right_on='Country',    # your dataframe
    how='left'
)
```

```python
world['Attacks'] = world['Attacks'].fillna(0)
```

```python
from libpysal.weights import Queen
```

```python
w_country = Queen.from_dataframe(world)
w_country.transform = 'R'   # row-standardization
```

```python
from esda.moran import Moran_Local
import numpy as np
```

```python
lisa = Moran_Local(y, w_country)
```

```python
world['local_I'] = lisa.Is
world['local_p'] = lisa.p_sim      # permutation p-values
world['local_q'] = lisa.q          # quadrant classification
```

```python
world['significant'] = world['local_p'] < 0.05
```

```python
def classify(row):
    if not row['significant']:
        return 'Not Significant'
    elif row['local_q'] == 1:
        return 'High-High'
    elif row['local_q'] == 2:
        return 'Low-High'
    elif row['local_q'] == 3:
        return 'Low-Low'
    elif row['local_q'] == 4:
        return 'High-Low'

world['cluster'] = world.apply(classify, axis=1)
```

```python
ols_train = sm.OLS(y_train, X_train).fit()
print(ols_train.summary())
```

```python
y_pred = ols_train.predict(X_test)
```
