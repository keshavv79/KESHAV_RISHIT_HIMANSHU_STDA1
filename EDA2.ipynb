<h1>Decomposition

```python
from statsmodels.tsa.seasonal import seasonal_decompose

decomp = seasonal_decompose(ts_attacks_m, model='additive', period=12)

decomp.plot()
plt.show()
```

```python
from statsmodels.tsa.stattools import adfuller

print("ADF original:", adfuller(ts_attacks_m)[1])

ts_diff = ts_attacks_m.diff().dropna()
print("ADF diff:", adfuller(ts_diff)[1])
```

<h1>Model

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))
plot_acf(ts_diff, lags=40)
plt.title("ACF - Differenced Series")
plt.show()

plt.figure(figsize=(12,4))
plot_pacf(ts_diff, lags=40)
plt.title("PACF - Differenced Series")
plt.show()
```

```python
train = ts_attacks_m[:'2014-12-01']
test = ts_attacks_m['2015-01-01':]
```

```python
from statsmodels.tsa.arima.model import ARIMA

model = ARIMA(train, order=(0,1,1))
model_fit = model.fit()

print(model_fit.summary())
```

```python
forecast = model_fit.forecast(steps=len(test))
```

```python
from sklearn.metrics import mean_squared_error
import numpy as np

rmse_raw = np.sqrt(mean_squared_error(test, forecast))
print("RMSE(RAW):", rmse_raw)
```

```python
residuals = model_fit.resid

plt.figure(figsize=(12,4))
plt.plot(residuals)
plt.title("Residuals - Raw Model")
plt.show()

plot_acf(residuals, lags=40)
plt.title("Residual ACF - Raw Model")
plt.show()
```

```python
from statsmodels.stats.diagnostic import acorr_ljungbox
print(acorr_ljungbox(residuals, lags=[10,20], return_df=True))
```

```python
ts_log = np.log(ts_attacks_m + 1)
```

```python
print("ADF log original:", adfuller(ts_log)[1])

ts_log_diff = ts_log.diff().dropna()
print("ADF log diff:", adfuller(ts_log_diff)[1])
```

```python
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
import matplotlib.pyplot as plt

plt.figure(figsize=(12,4))
plot_acf(ts_log_diff, lags=40)
plt.title("ACF - Log Differenced")
plt.show()

plt.figure(figsize=(12,4))
plot_pacf(ts_log_diff, lags=40)
plt.title("PACF - Log Differenced")
plt.show()
```

```python
train_log = ts_log[:'2014-12-01']
test_log = ts_log['2015-01-01':]
```

```python
from statsmodels.tsa.arima.model import ARIMA

model_log = ARIMA(train_log, order=(0,1,1))
model_log_fit = model_log.fit()

print(model_log_fit.summary())
```

```python
forecast_log = model_log_fit.forecast(steps=len(test_log))
```

```python
forecast_original = np.exp(forecast_log) - 1
```

```python
from sklearn.metrics import mean_squared_error

import numpy as np

rmse_log_model = np.sqrt(
    mean_squared_error(test, forecast_original)
)

print("RMSE (Log Model):", rmse_log_model)
```

```python
residuals_log = model_log_fit.resid

plt.figure(figsize=(12,4))
plt.plot(residuals_log)
plt.title("Residuals - Log Model")
plt.show()

plot_acf(residuals_log, lags=40)
plt.title("ACF Residuals - Log Model")
plt.show()
```

```python
from statsmodels.stats.diagnostic import acorr_ljungbox

acorr_ljungbox(residuals_log, lags=[10,20], return_df=True)
```

```python
plt.figure(figsize=(12,5))

# Plot training data
plt.plot(train.index, train, label="Training Data")

# Plot actual test data
plt.plot(test.index, test, label="Actual (Test)")

# Plot forecast
plt.plot(test.index, forecast_original, 
         label="Forecast (Log ARIMA)", 
         linestyle='--')

plt.title("Actual vs Forecast - Monthly Attack Count")
plt.xlabel("Year")
plt.ylabel("Number of Attacks")
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,5))

plt.plot(test.index, test, label="Actual")
plt.plot(test.index, forecast_original, 
         label="Forecast", linestyle='--')

plt.title("Forecast vs Actual (2015–2017)")
plt.legend()
plt.grid(True)
plt.show()
```

```python
from statsmodels.tsa.arima.model import ARIMA
from sklearn.metrics import mean_squared_error
import numpy as np

# Fit ARIMA(1,1,1) on log-transformed series
model_111 = ARIMA(train_log, order=(1,1,1))
model_111_fit = model_111.fit()

print(model_111_fit.summary())
```

```python
forecast_111_log = model_111_fit.forecast(steps=len(test_log))

# Back-transform
forecast_111_original = np.exp(forecast_111_log) - 1
```

```python
rmse_111 = np.sqrt(
    mean_squared_error(test, forecast_111_original)
)

print("RMSE (ARIMA 1,1,1 - Log):", rmse_111)
print("AIC (ARIMA 1,1,1 - Log):", model_111_fit.aic)
```

```python
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt

resid_111 = model_111_fit.resid

plt.figure(figsize=(12,4))
plot_acf(resid_111, lags=40)
plt.title("ACF Residuals - ARIMA(1,1,1)")
plt.show()
```

<h1>LR

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error

# Training and test sets (same split as ARIMA)
train = ts_attacks_m[:'2014-12-01']
test = ts_attacks_m['2015-01-01':]

# Create numeric time variable
train_time = np.arange(len(train)).reshape(-1,1)
test_time = np.arange(len(train), len(train)+len(test)).reshape(-1,1)
```

```python
lr = LinearRegression()
lr.fit(train_time, train.values)

forecast_lr = lr.predict(test_time)
```

```python
rmse_lr = np.sqrt(mean_squared_error(test, forecast_lr))

print("Linear Regression RMSE:", rmse_lr)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(test.index, test, label="Actual")
plt.plot(test.index, forecast_lr, 
         label="Linear Regression Forecast", linestyle='--')

plt.title("Linear Regression Forecast vs Actual (2015–2017)")
plt.legend()
plt.grid(True)
plt.show()
```
